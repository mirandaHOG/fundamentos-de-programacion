```python
tabla = int(input("Dame la tabla de multiplicar que deseas"))
inicio = int(input("Dame el inicio del rango"))
fin = int(input("Dame el fin del rango"))
aumento = int(input("Dame el aumento del rango"))
while inicio >= fin:
  print("el limite tiene que ser mayor al inicio")
  fin = int(input("Dame el fin del rango"))

while inicio <= fin:
  print(tabla, "*", inicio, "=", tabla*inicio)
  inicio = inicio + aumento
```
