```python
# ============================================
# SISTEMA DE PRÉSTAMOS
# Versión 1 - Programación basica y uso de listas
# ============================================

# Lista de alumnos activos en el sistema
import datetime as dt
alumnos = [
    ["26040200", "Juan", "Sistemas Coputacionales"],
    ["26040201", "Rosa", "Biomedicina"],
    ["26040202", "Alba", "Arquitectura"],
    ["26040203", "Martines", "Ing. Civil"],
    ["26040204", "Mendez", "Ing. en Inteligencia Artificial"],
    ["26040205", "Alexa", "Ing. Mecatronica"],
    ["26040206", "Amy", "Biomediciona"],
    ["26040207", "Marieli", "Arquitectura"],
    ["26040208", "Isabela", "Sistemas Computacionales"],
    ["26040209", "Marilin", "Ing.Civil"]
    ]
prestamos = [
    ["26040200", "Python, Java, Vs Code", "LC1", "PC-LC1-02"],
    ["26040201", "AutoCAD, MATLAB", "LC1", "PC-LC1-03"],
    ["26040202", "Python, Java, Vs Code", "LC2", "PC-LC2-02"],
    ["26040203", "AutoCAD, MATLAB", "LC2", "PC-LC2-03"],
    ["26040204", "Python, Java, VS Code", "LC3", "PC-LC3-02"],
    ["26040205", "AutoCAD, MATLAB", "LC3", "PC-LC3-03"],
]
# Cada valor en la lista de computadoras tiene su lista correspondiente con el
# aula donde esta ubicada, el software instalado y el estado de esa computadora
computadoras = [
                ["PC-LC1-01", "LC1", "Python, VS Code", "Disponible"],
                ["PC-LC1-02", "LC1", "Python, Java, VS Code", "Prestada"],
                ["PC-LC1-03", "LC1", "AutoCAD, MATLAB", "Reservada"],
                ["PC-LC2-01", "LC2", "Python, VS Code", "Disponible"],
                ["PC-LC2-02", "LC2", "Python, Java, VS Code", "Prestada"],
                ["PC-LC2-03", "LC2", "AutoCAD, MATLAB", "Reservada"],
                ["PC-LC3-01", "LC3", "Python, VS Code", "Disponible"],
                ["PC-LC3-02", "LC3", "Python, Java, VS Code", "Prestada"],
                ["PC-LC3-03", "LC3", "AutoCAD, MATLAB", "Reservada"],
                ["PC-LC3-04", "LC3", "Python, VS Code", "Mantenimiento"],
                ]
necesidades = [
                ["26040200", "Android Studio", "2026-10-14 19:16:34"]
]
def listar_computadoras():
    print("\n--- COMPUTADORAS ---")
    for i in range(len(computadoras)):
        print("Aula:", computadoras[i][1], "ID:", computadoras[i][0], "Software:", computadoras[i][2], "Estado:", computadoras[i][3])

def buscar_software():
    num_control = input("\nIngrese el número de control: ")
    programa = input("\nIngrese el software que desea buscar: ")
    print("\n--- RESULTADOS ---")
    encontrado = False
    for i in range(len(computadoras)):
        if programa.lower() in computadoras[i][2].lower():
                print("ID:", computadoras[i][0], "Aula:", computadoras[i][1], "Software:", computadoras[i][2], "Estado:", computadoras[i][3])
                encontrado = True

    if encontrado == False:
        print("No se encontraron computadoras con ese software.")
        necesidades.append([num_control, programa, dt.datetime.now()])

def solicitar_prestamo():
    print("\n--- SOLICITAR PRÉSTAMO ---")
    num_control = input("\nIngrese el número de control del alumno: ")
    if num_control not in alumnos[0]:
        print("El alumno no existe.")
        return
    id_computadora = input("Ingrese el ID de la computadora: ")
    encontrado = False
    for i in range(len(computadoras)):
        if computadoras[i][0] == id_computadora:
            encontrado = True
            if computadoras[i][3] == "Disponible":
                software = input("Ingrese el nombre del software a utilizar: ").strip()

                # Verificación de la existencia del software
                if software in computadoras[i][2]:
                    computadoras[i][3] = "Prestada"
                    print("\n¡Préstamo registrado correctamente!")
                    print("Computadora:", computadoras[i][0])
                    print("Aula:", computadoras[i][1])
                    print("Software:", software)
                    prestamos.append([num_control, computadoras[i][2], computadoras[i][1], computadoras[i][0]])
                else:
                    necesidades.append([num_control, software, dt.datetime.now()])
                    print(f"\nEl software '{software}' no está disponible o no existe.")
                    print("Se ha registrado en la lista de solicitudes no encontradas.")
            else:
                print("La computadora no está disponible.")
            break
    if not encontrado:
        print("La computadora no existe.")

def Devolver_computadora():
  print("\n--- DEVOLVER COMPUTADORA ---")
  id_computadora = input("dame el ID de la computadora: ")
  existe = False
  for i in range(len(computadoras)):
      if computadoras[i][0] == id_computadora:
          existe = True
          if computadoras[i][3] == "Prestada" or computadoras[i][3] == "Reservada":
                computadoras[i][3] = "Disponible"
                print("la computadora", id_computadora, "ya esta lista")
          else:
                print("la computadora", id_computadora, "no esta prestada/reservada")
          if existe == False:
            print("la computadora no existe", id_computadora)

def crear_reservacion():
  print("\n--- CREAR RESERVACION ---")
  num_control = input("\nIngrese el número de control: ")
  encontrado1 = False
  if num_control in alumnos:
    encontrado1 = True
    id_computadora = input("Ingrese el ID de la computadora: ")
    encontrado2 = False
    for i in range(len(computadoras)):
        if computadoras[i][0] == id_computadora:
            encontrado2 = True
            if computadoras[i][3] == "Disponible":
                computadoras[i][3] = "Reservada"
                print("Computadora reservada correctamente.")
                print("Computadora:", computadoras[i])
                print("Aula:", computadoras[i][1])
                prestamos.append([num_control, computadoras[i][2], computadoras[i][1], computadoras[i][0]])
            else:
                print("La computadora no está disponible.")
            if encontrado2 == False:
                print("La computadora no existe.")
        if encontrado1 == False:
            print("El alumno no existe.")

def Agregar_computadora():
  print("\n--- AGREGAR COMPUTADORA ---")
  id_computadora = input("ID de la computadora: ")
  aula = input("Aula: ")
  programas = input("Software instalado: ")
  estado = input("Estado: ")
  nueva_computadora = [id_computadora, aula, programas, estado]
  computadoras.append(nueva_computadora)
  print("\nComputadora agregada correctamente.")

def Historial_prestamos():
  print("\n--- Historial de Prestamos ---")
  for i in range(len(prestamos)):
     print("numero de control:", prestamos[i][0], "ID:", prestamos[i][3], "Software:", prestamos[i][1], "Aula:", prestamos[i][2])

def historial_necesidades():
  print("\n--- Historial de necesidades ---")
  for i in range(len(necesidades)):
    print("numero de control:", necesidades[i][0], "software:", necesidades[i][1], "fecha:", necesidades[i][2])

def mostrar_estadisticas():
    print("\n======= Estadísticas por programa ===========")
    conteo_programas = {}
    for prestamo in prestamos:
        prog = prestamo[1].strip()
        conteo_programas[prog] = conteo_programas.get(prog, 0) + 1

    for prog, veces in conteo_programas.items():
        print(f"{prog.lower()} : {veces} veces")

    print("\n======= Estadísticas por carrera ===========")
    conteo_carreras = {}
    for prestamo in prestamos:
        num_ctrl_prestamo = prestamo[0]
        carrera_alumno = "Desconocida"
        for alumno in alumnos:
            if alumno[0] == num_ctrl_prestamo:
                carrera_alumno = alumno[2]
                break
        conteo_carreras[carrera_alumno] = conteo_carreras.get(carrera_alumno, 0) + 1

    for carrera, veces in conteo_carreras.items():
        print(f"{carrera}: {veces} veces")

# MENÚ PRINCIPAL
# Usamos un ciclo While que se repetira indefinidamente hasta que el usuario
# teclee la opcion 7 (salir) o cualquier numero superior a 7
opcion = 0
while opcion <= 9:
    print("\n=================================")
    print("      SISTEMA DE PRÉSTAMOS")
    print("=================================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar préstamo")
    print("4. Devolver computadora")
    print("5. Crear reservación")
    print("6. Agregar computadora")
    print("7. Historial de prstamos")
    print("8. Historial de necesidades")
    print("9. Estadisticas")
    print("10. Salir")
    print("=================================")
    opcion = int(input("Seleccione una opción: "))

    # 1. Consultar computadoras
    # si el usuario selecciono la opcion 1 se hace un ciclo (for) que recorre
    # toda la lista de computadoras e imprime los valores de las demás listas
    if opcion == 1:
      listar_computadoras()
    # 2. Buscar por software
    elif opcion == 2:
      buscar_software()
    # 3. Solicitar préstamo
    elif opcion == 3:
        solicitar_prestamo()
    # 4. Devolver computadora
    elif opcion == 4:
        Devolver_computadora()
    elif opcion == 5:
        crear_reservacion()
    # 6. Agregar computadora
    elif opcion == 6:
        Agregar_computadora()
    elif opcion == 7:
        Historial_prestamos()
    elif opcion == 8:
        historial_necesidades()
    elif opcion == 9:
        mostrar_estadisticas()
```
