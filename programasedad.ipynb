```python
edad = int(input("Dame tu edad"))
if edad <= 10:
    print("Eres un niño")
elif edad <= 18:
    print("Eres un adolescente")
elif edad <= 30:
    print("Eres un joven adulto")
elif edad <= 65:
    print("Eres un adulto")
elif edad <= 100:
    print("Eres un anciano")
else:
    print("Edad incorrecta")
```
